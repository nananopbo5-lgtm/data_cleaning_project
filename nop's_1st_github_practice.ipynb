```python
# now we need to replace these ERROR and UNKNOWN values across our data with NaN
df = df.replace(['ERROR', 'UNKNOWN'], np.nan) # to access Nan we code NumPy ie np.nan

df.head(10)
```

```python
# now we will convert it into a Right DATA TYPES. AKA -> changing data types
# .astypes helps you convert a columns data type.
# NaN values cannot be contain in INT so we use Int64 which means higher position int numbers.
df['Quantity'] = df['Quantity'].astype('Int64') 
```

```python
# dropping records where Transaction Data is missing
df.dropna(subset = ['Transaction Date'])
# here we can check at what percent our data is treated. if any person left we will re do it
```

```python
# dropping records where Transaction Data is missing by adding df = in first
df = df.dropna(subset = ['Transaction Date'])
```

```python
df.info() # all records are shown here
```

```python
# now we will see the Payment Method and Location which we cannot delete directly 
# either we can permanenetly drop these columns and save ourself from trouble or we can create a flag(as an INA 'Infromation Not Available)
# That way we can have entire 9540 records and we'll be able to say in the payment method and location that if NaN values are there we FLAG it as INA.
```

```python
# replacing missing values in Payment Method and Location with INA(information not available)
# reason --> dropping there high volumn records would lead to significant data loss
# and inputing INA avoids bias
df[['Payment Method', 'Location']].fillna('INA')
```

```python
# this just we do over write the code
df[['Payment Method', 'Location']] = df[['Payment Method', 'Location']].fillna('INA')
```

```python
# now we will see the data with something i.e Mean Median or Mode.
# Mean and Median wont work in str data type. these operation are for no's only
# but Mode(most commonly accuring value) will work here.
# but the problem is that 30%/40% data are artificially generated by us which can create BIAS.
# It might look that 70% of transaction data made from UPI or 70% transaction are done in Store that means food fall is high
# But that might not be true coz we ourself filled the 30%-40% data.
# so not we can't delete it nor can fill it artificially from MODE option.
# so the only option left with us is 'FLAG' like 'INA'
# that means data is with us, artificial bias won't be created and we will have the records where the information is unavailable.
```

```python
# now if we run the code we will see the deleting values which earlier present in diff values.
df.isna().sum() * 100 / df.shape[0]
```

```python
# now we can see the ITEM, QUANTITY, PRICE P/U, TOTAL SPENT
# we'll go with lowest first i.e QUANTITY
# quantity is a number so we can take MEAN, MEDIAN or MODE.
# so will go with MEAN and MEDIAN
# if both result same no then we can choose either.
# if diff result then choose median coz median doesn't care about outlier.

# i.e; inspecting mean and median values for Quantity.
quantity_mean, quantity_median = round(df['Quantity'].mean(), 0), round(df['Quantity'].median(), 0)

print(f'Mean Quantity: {quantity_mean}\nMedian Quantity: {quantity_median}')
```

```python
# since mean and median values are similar, replacing missing value in quantity with mean value
df['Quantity'] = df['Quantity'].fillna(quantity_mean)
```

```python
# now we can see the quantity column is taken over/care by 0
df.isna().sum() * 100 / df.shape[0]
```

```python
# now we are left with ITEM = 9.716, PRICE P/U = 5.303, TOTAL SPENT = 4.989
# we'll go with replacing missing values in ITEM(item is text based in nature) with MODE value
# in text value only MODE values work but not mean and median.
df['Item'].mode()
```

```python
# the above juice gave us 0 index so we need
df['Item'].mode()[0]
```

```python
# now
df['Item'] = df['Item'].fillna(df['Item'].mode()[0])
```

```python
# now we check data of Item which is taken care of by replacing 9.7 to 0.
df.isna().sum() * 100 / df.shape[0]
```

```python
# now we will go for PRICE P/U which is 5.303
# but before that we have to know price p/u will be diff for each cafes menu
# we cannot use avg, mean, median, mode it will throw error
# so we'll go with GroupBy Operation. --> means category wise.

# first we need to check how many values exist inside of rice p/u data.
# we do checking how many distinct prices exist for each Item category and how many times each price point is registered.
df.groupby('Item')['Price Per Unit'].value_counts()
```

```python
# by getting output value i.e; item per unit which is diff from each item
# now do you think we can calculate mean median and mode here? what does more sense?
# for first iten i.e CAKE its price p/u is always 3.0 which is always be same mean median and mode. and same for each items which have same price per unit.
# but for juice which have diff price per unit which is 3,4,2,5,1,1 whose avg/mean/median/mode will be 3 only.
# its means we can go for Median or Mode in our case.

# so we do filling missing price per unit values with the most frequent (Mode) price for each category
df.groupby('Item')['Price Per Unit'].transform(lambda x: x.fillna(x.mode()[0]))
```

```python
# we'll over write the code inside of price P/U
df['Price Per Unit'] = df.groupby('Item')['Price Per Unit'].transform(lambda x: x.fillna(x.mode()[0]))
```

```python
# now we check for Price P/U is taken care of by 0
df.isna().sum() * 100 / df.shape[0]
```

```python
# so now the only missing column is left with Total spent which is 4.989.
# Total spent is a no so we always calculated as the amt*quantity that you have ordered.

# so we do filling missing records for total spent by calculating the product of quantity times price per unit.
df['Total Spent'] = df['Total Spent'].fillna(df['Quantity'] * df['Price Per Unit'])
```

```python
# now we check Total Spent is taken care of by 0
df.isna().sum() * 100 / df.shape[0]
```

```python
# now we do storing clean data
# we will remove the index present in the data
df.to_csv('/workspaces/data_cleaning_project/data/clean/Cafe Sales (Clean).csv', index = False)
```
